**Method:**  
The method used to find the answer to this question starts with a steel wire that will be stretched between two fixed supports. The wire will be plucked so that it vibrates, and its fundamental natural frequency will be determined from the measured signal on the oscilloscope using a Fourier transform. 
    The first measurements will be taken at room temperature to determine how the natural frequency depends on the tension of the wire.
    The temperature of the wire will be increased by passing a DC current through it. Due to electrical resistance, the wire will warm up. The current will be varied to obtain wire temperatures between approximately 15 C and 50 C. At several points in this range, the fundamental natural frequency will be measured.
