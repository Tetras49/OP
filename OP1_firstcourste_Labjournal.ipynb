# Processing
**Description of processing of raw data into scientific evidence:**


```python



files = [
    (25, "Metingen_temperaturen/25.csv"),
    (32, "Metingen_temperaturen/32.csv"),
    (34, "Metingen_temperaturen/34.csv"),
    (40, "Metingen_temperaturen/40.csv"),
    (49, "Metingen_temperaturen/49.csv"),
    (50, "Metingen_temperaturen/50.csv"),
    (55, "Metingen_temperaturen/55.csv"),
    (55, "Metingen_temperaturen/55 (2).csv"),
    (60, "Metingen_temperaturen/60.csv"),
]


def damped_sine(t, A, gamma, f, phi, C):
    return A * np.exp(-gamma * t) * np.sin(2*np.pi*f*t + phi) + C


temperatures = []
frequencies = []
frequency_errors = []


for temperature, filename in files:

    df = pd.read_csv(filename, skiprows=12, names=["Time", "Voltage"])

    time = df["Time"].to_numpy()
    voltage = df["Voltage"].to_numpy()

    # make time start at 0
    time = time - time[0]

    voltage_centered = voltage - np.mean(voltage)

    dt = np.mean(np.diff(time))

    # FFT for initial estimate
    fft = np.fft.rfft(voltage_centered)
    freq = np.fft.rfftfreq(len(voltage_centered), d=dt)
    amplitude = np.abs(fft)

    peak = np.argmax(amplitude[1:]) + 1

    y1 = amplitude[peak - 1]
    y2 = amplitude[peak]
    y3 = amplitude[peak + 1]

    p = 0.5 * (y1 - y3) / (y1 - 2*y2 + y3)

    f_guess = freq[peak] + p * (freq[1] - freq[0])


    # initial guesses for curve fit
    A_guess = (np.max(voltage) - np.min(voltage)) / 2
    gamma_guess = 10
    phi_guess = 0
    C_guess = np.mean(voltage)

    popt, pcov = curve_fit(
        damped_sine,
        time,
        voltage,
        p0=[A_guess, gamma_guess, f_guess, phi_guess, C_guess],
        maxfev=20000
    )

    # fitted parameters
    A_fit, gamma_fit, f_fit, phi_fit, C_fit = popt

    # uncertainties from covariance matrix
    perr = np.sqrt(np.diag(pcov))

    f_error = perr[2]

    temperatures.append(temperature)
    frequencies.append(f_fit)
    frequency_errors.append(f_error)

    print(
        temperature, "°C:",
        f_fit, "±", f_error, "Hz"
    )


# temperature uncertainty
temperature_error = 3


# Linear fit
fit = np.polyfit(temperatures, frequencies, 1)

slope = fit[0]
intercept = fit[1]

T_fit = np.linspace(min(temperatures), max(temperatures), 100)
f_fit_line = slope * T_fit + intercept


# Plot
plt.errorbar(
    temperatures,
    frequencies,
    xerr=temperature_error,
    yerr=frequency_errors,
    fmt="o",
    capsize=4,
    label="Measurements"
)

plt.plot(T_fit, f_fit_line, label="Linear fit")

plt.xlabel("Temperature (°C)")
plt.ylabel("Natural frequency (Hz)")
plt.title("Natural frequency vs temperature")

plt.grid()
plt.legend()

plt.show()
print(frequency_errors)
print("Slope =", slope, "Hz/°C")
```

**New research question:** Can measurements of the fundamental natural frequency be used to determine the change in linear mass density due to a change in temperature in a range from 25 to 60 C?

```python

def calculate_theoretical_frequency(
    temp_c, 
    L=0.2341,        # Vibrating length (m) - adjusted to match your 539.5 Hz baseline at 25°C
    d0=0.5e-3,       # Initial wire diameter (m)
    T_tension=100.0, # Constant tension (N)
    rho0=8000.0,     # Density at 25°C (kg/m^3)
    alpha=16e-6,     # Linear thermal expansion coefficient (/°C)
    E0=193e9,        # Young's Modulus at 25°C (Pa)
    dE_dT=-75e6,     # Change in Young's Modulus per °C (Pa/°C)
    T0=25.0          # Baseline reference temperature (°C)
):

    temp_c = np.asarray(temp_c)
    delta_T = temp_c - T0
    
 
    A0 = np.pi * (d0 / 2)**2
    mu0 = rho0 * A0

    A_T = A0 * (1 + 2 * alpha * delta_T)
    E_T = E0 + dE_dT * delta_T
    

    strain_T = T_tension / (A_T * E_T)

    mu_T = mu0 / ((1 + alpha * delta_T) * (1 + strain_T))
    

    f_T = (1 / (2 * L)) * np.sqrt(T_tension / mu_T)
    
    return f_T




theory_temps = np.linspace(25, 60, 100)
theory_freqs = calculate_theoretical_frequency(theory_temps)


plt.figure(figsize=(9, 5), dpi=100)
plt.plot(theory_temps, theory_freqs, label="Theory (Constant Tension $T=100\\text{N}$)", 
    color="crimson", linewidth=2)
plt.title("Fundamental Frequency vs Temperature of 316 Stainless Steel Wire", fontsize=12)
plt.xlabel("Temperature (°C)", fontsize=11)
plt.ylabel("Fundamental Frequency (Hz)", fontsize=11)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend(fontsize=10)
plt.tight_layout()
plt.show()
```
